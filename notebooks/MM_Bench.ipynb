```python
from datasets import load_dataset
import random
from IPython.display import display, Image
from io import BytesIO
from PIL import Image as PILImage

ds = load_dataset("Lin-Chen/MMStar")
```

```python
ds['val']
```

```python
# Explore the dataset
val_dataset = ds['val']

# Display basic information about the dataset
print("Features:", val_dataset.features)
print("Number of rows:", val_dataset.num_rows)

# Display the first few rows of the dataset
print("\nFirst few rows:")
print(val_dataset[:5])

# Display statistics about the dataset
print("\nStatistics:")
print(val_dataset.info)

# Explore unique categories and l2_categories
print("\nUnique categories:")
print(set(val_dataset['category']))

print("\nUnique l2_categories:")
print(set(val_dataset['l2_category']))

# Display a sample of meta_info for further exploration
print("\nSample meta_info:")
print(val_dataset['meta_info'][:5])
```

```python

num_examples = 10
random_indices = random.sample(range(val_dataset.num_rows), num_examples)
examples = val_dataset.select(random_indices)

for i, example in enumerate(examples):
    image = example['image']  # Adjust if the image is not a path
    question = example['question']
    answer = example['answer']
    
    # print(image_data)
    # Display the image
    # display(Image(filename=image_path))
    # Convert the byte data to an image
    # pil_image = PILImage.open(BytesIO(image_data))
    display(image)
    
    # Display the question and answer
    print(f"Question: {question}\nAnswer: {answer}\n")

```
