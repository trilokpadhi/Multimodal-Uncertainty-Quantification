```python
import re

def generate_premise(question, options, correct_answer):
    # Extract the question type (What, Where, How many, etc.)
    question_type = question.split()[0].lower()
    
    # Extract the main subject of the question
    subject = re.search(r'(?:is|are) (?:the |a )?(.*?) (?:in|of|on|at)', question, re.IGNORECASE)
    subject = subject.group(1) if subject else "subject"
    
    # Get the correct answer text
    answer_text = options[correct_answer].split(' - ')[1] if ' - ' in options[correct_answer] else options[correct_answer]
    
    # Generate premise based on question type
    if question_type == 'what':
        premise = f"The {subject} is {answer_text}."
    elif question_type == 'where':
        premise = f"The {subject} is located {answer_text}."
    elif question_type == 'how':
        if 'many' in question.lower():
            premise = f"There are {answer_text} {subject}."
        else:
            premise = f"The {subject} is {answer_text}."
    elif question_type == 'who':
        premise = f"{answer_text} is the {subject}."
    elif question_type == 'when':
        premise = f"The {subject} occurs {answer_text}."
    else:
        premise = f"The {subject} is {answer_text}."
    
    return premise

# Example usage:
question = "What is the colour of car in the image?"
options = {"a": "a - red", "b": "b - blue", "c": "c - green", "d": "d - yellow"}
correct_answer = "b"

premise = generate_premise(question, options, correct_answer)
print(premise)
```
