### Scene Graphs with Grounded SAM

```python
import json
import networkx as nx
import matplotlib.pyplot as plt

# Load the JSON data from file
with open('/home/ubuntu/Multimodal-Uncertainty-Quantification/playground/scene_graph/SceneGraphGenZeroShotWithGSAM/outputs/scenegraph.json', 'r') as f:
    data = json.load(f)
```

```python
# Create a directed graph
G = nx.DiGraph()

# Add nodes with attributes
for node in data['nodeDataArray']:
    G.add_node(node['key'], color=node['color'])

# Add edges
for edge in data['linkDataArray']:
    G.add_edge(edge['from'], edge['to'])

# Draw the graph
pos = nx.spring_layout(G)
node_colors = [data['color'] for _, data in G.nodes(data=True)]
labels = {node: node for node in G.nodes()}

nx.draw(G, pos, labels=labels, node_color=node_colors, with_labels=True, arrows=True)
plt.show()
```
