```python
# Ramneets Method, based on grounding models confidence, 
# i have to check the agreement between the grounding models, lets say i take llama3.2v 11b and llama3.2v 70b,
# agrement, is how many times they agree on the grounding, both are list of yes, so i have to see how many times they agree on the yes or no, 

def calculate_agreement(response_list_1, response_list_2):
    if not response_list_1 or pd.isna(response_list_1) or not response_list_2 or pd.isna(response_list_2):
        return 0
    responses_1 = ast.literal_eval(response_list_1)
    responses_2 = ast.literal_eval(response_list_2)
    agreement = sum([1 for r1, r2 in zip(responses_1, responses_2) if r1 == r2])
    total_count = len(responses_1)
    return agreement / total_count if total_count > 0 else 0

# Apply the calculate_agreement function to each row
merged_df['grounding_llama32_11b_llama32_70b_agreement'] = merged_df.apply(
    lambda row: calculate_agreement(row['grounding_llama32_11b_processed'], row['grounding_llama32_70b_processed']),
    axis=1
)

```

```python
import math

def item_level_entropy_3lists(list1, list2, list3):
    """
    For each of the 20 positions, count how many 'yes' among the 3 lists.
    Compute binary entropy of that mini-distribution (p_yes, p_no).
    Then average across the 20 items.
    """
    n = len(list1)  # assume each list has the same length, e.g. 20
    total_entropy = 0.0
    
    for i in range(n):
        # yes_count for item i across the 3 lists
        yes_count = sum(lst[i].lower() == 'yes' for lst in [list1, list2, list3])
        p_i = yes_count / 3.0
        
        if p_i == 0 or p_i == 1:
            H_i = 0.0
        else:
            H_i = - (p_i * math.log2(p_i) + (1 - p_i) * math.log2(1 - p_i))
        
        total_entropy += H_i
    
    # Average entropy across the 20 items
    return total_entropy / n

import math

def item_level_entropy_2experts(expert1_responses, expert2_responses):
    """
    Computes average binary entropy across items, where each item has
    'yes'/'no' from two experts.

    - expert1_responses: list of length N, each "yes" or "no"
    - expert2_responses: list of length N, each "yes" or "no"

    Returns a float: average binary entropy across the N items.

    For each item i:
      - p_i = (#yes among 2 experts) / 2
      - H_i = -[p_i log2(p_i) + (1 - p_i) log2(1 - p_i)] (0 if p_i=0 or 1)
    Then we average H_i over all items.
    """
    n = len(expert1_responses)
    total_entropy = 0.0
    
    for i in range(n):
        yes_count = 0
        if expert1_responses[i].lower() == "yes":
            yes_count += 1
        if expert2_responses[i].lower() == "yes":
            yes_count += 1
        
        p_i = yes_count / 2.0
        
        if p_i == 0 or p_i == 1:
            H_i = 0.0
        else:
            H_i = -(p_i * math.log2(p_i) + (1 - p_i) * math.log2(1 - p_i))
        
        total_entropy += H_i
    
    avg_entropy = total_entropy / n
    return avg_entropy
def item_level_entropy_3experts(expert1_responses, expert2_responses, expert3_responses):
    """
    Compute the average binary entropy across items, where each item has 3 'yes'/'no' votes
    from three experts.
    
    - expert1_responses: list of length N (e.g., 20), each either "yes" or "no"
    - expert2_responses: list of length N, each either "yes" or "no"
    - expert3_responses: list of length N, each either "yes" or "no"
    
    Returns the average binary entropy: 
        - For each item i, we compute the fraction p_i = (#yes among the 3 experts) / 3
        - Then we compute the binary entropy H_i = -[p_i log2(p_i) + (1-p_i) log2(1-p_i)]
          (with 0 if p_i=0 or 1)
        - Finally, we average H_i across all items.
    """
    # Assume all lists have the same length
    n = len(expert1_responses)
    total_entropy = 0.0
    
    for i in range(n):
        # Count how many experts said "yes"
        yes_count = 0
        if expert1_responses[i].lower() == "yes":
            yes_count += 1
        if expert2_responses[i].lower() == "yes":
            yes_count += 1
        if expert3_responses[i].lower() == "yes":
            yes_count += 1

        # Fraction that said "yes"
        p_i = yes_count / 3.0
        
        # Compute binary entropy
        if p_i == 0 or p_i == 1:
            H_i = 0.0
        else:
            H_i = -(p_i * math.log2(p_i) + (1 - p_i) * math.log2(1 - p_i))
        
        total_entropy += H_i
    
    # Average entropy across all items
    avg_entropy = total_entropy / n
    return avg_entropy

# Apply the item_level_entropy_3lists function to 'grounding_llama32_11b_processed','grounding_llama32_70b_processed', 'grounding_qwen_vl_processed',

merged_df['grounding_models_entropy_llama11b_llama70b_qwenvl'] = merged_df.apply( lambda row: item_level_entropy_3lists(ast.literal_eval(row['grounding_llama32_11b_processed']), ast.literal_eval(row['grounding_llama32_70b_processed']), ast.literal_eval(row['grounding_qwen_vl_processed'])), axis=1)
# merged_df['grounding_models_entropy_llama11b_llama70b_qwenvl'].head(
    
# now caclaute the entropy between llama32_11b and llama32_70b
merged_df['grounding_models_entropy_llama11b_llama70b'] = merged_df.apply( lambda row: item_level_entropy_2experts(ast.literal_eval(row['grounding_llama32_11b_processed']), ast.literal_eval(row['grounding_llama32_70b_processed'])), axis=1)

# now caclaute the entropy between llama32_11b and qwen_vl
merged_df['grounding_models_entropy_llama11b_qwenvl'] = merged_df.apply( lambda row: item_level_entropy_2experts(ast.literal_eval(row['grounding_llama32_11b_processed']), ast.literal_eval(row['grounding_qwen_vl_processed'])), axis=1)

# now caclaute the entropy between llama32_70b and qwen_vl
merged_df['grounding_models_entropy_llama70b_qwenvl'] = merged_df.apply( lambda row: item_level_entropy_2experts(ast.literal_eval(row['grounding_llama32_70b_processed']), ast.literal_eval(row['grounding_qwen_vl_processed'])), axis=1)
```

```python
import torch

def are_semantically_equivalent(respA, respB, question, tokenizer, model, device='cpu'):
    """
    Checks if respA and respB are mutually entailing given the question context.
    Returns True if label=2 in both directions, else False.
    """

    premise_forward  = f"{respA}"
    hypothesis_fwd   = f"{respB}"

    premise_backward = f"{respB}"
    hypothesis_bwd   = f"{respA}"

    # Forward pass
    inputs_f = tokenizer.encode_plus(
        premise_forward, hypothesis_fwd, 
        return_tensors='pt', truncation=True
    )
    inputs_f = {k: v.to(device) for k, v in inputs_f.items()}
    with torch.no_grad():
        logits_f = model(**inputs_f).logits
    label_f = torch.argmax(logits_f, dim=1).item()  # e.g. 0/1/2

    # Backward pass
    inputs_b = tokenizer.encode_plus(
        premise_backward, hypothesis_bwd, 
        return_tensors='pt', truncation=True
    )
    inputs_b = {k: v.to(device) for k, v in inputs_b.items()}
    with torch.no_grad():
        logits_b = model(**inputs_b).logits
    label_b = torch.argmax(logits_b, dim=1).item()

    # Check mutual entailment
    return (label_f == 2) and (label_b == 2)
```

```python
def cal_entropy_grounding_semantic(row, tokenizer, model, device='cpu'):
    responses = ast.literal_eval(row["responses"])
    groundings= ast.literal_eval(row["grounding_llama32_70b_processed"])
    question  = row["question"]  # if you stored it in the row
    n = len(responses)
    if n == 0 or len(groundings) != n:
        return 0.0

    # Build adjacency with mutual entailment
    adjacency = [[False]*n for _ in range(n)]
    for i in range(n):
        for j in range(i+1, n):
            # check if responses i and j are semantically equivalent
            if are_semantically_equivalent(
                   responses[i], responses[j], 
                   question, 
                   tokenizer, 
                   model, 
                   device=device
               ):
                adjacency[i][j] = True
                adjacency[j][i] = True

    # Then BFS cluster, same logic as before
    visited = [False]*n
    clusters = []
    for i in range(n):
        if not visited[i]:
            stack = [i]
            visited[i] = True
            cluster_members = []
            while stack:
                node = stack.pop()
                cluster_members.append(node)
                for nb in range(n):
                    if adjacency[node][nb] and not visited[nb]:
                        visited[nb] = True
                        stack.append(nb)
            clusters.append(cluster_members)

    # Compute grounding entropy
    grounding_score_with_entropy = []
    for c in clusters:
        cluster_size = len(c)
        yes_count = sum(1 for idx in c if groundings[idx].lower() == "yes")
        p = yes_count / cluster_size

        if p == 0 or p == 1:
            h_cluster = 0.0
        else:
            h_cluster = - (p * math.log2(p) + (1 - p)*math.log2(1 - p))

        # lets take the grounding score of the cluster, no of yes in the cluster / total no of responses in the cluster
        no_of_yes_in_cluster = sum(1 for idx in c if groundings[idx].lower() == "yes")
        cluster_size = len(c)
        grounding_score = no_of_yes_in_cluster / cluster_size
        grounding_score_of_the_cluster = grounding_score * h_cluster
        # weighted_sum += grounding_score_of_the_cluster
        grounding_score_with_entropy.append(grounding_score_of_the_cluster)
        
    return sum(grounding_score_with_entropy) / len(grounding_score_with_entropy)
```

```python
import math
import ast

def cal_entropy_grounding_semantic_debug(row, tokenizer, model, device='cpu'):
    """
    Computes grounding entropy by clustering responses based on semantic equivalence.
    This version prints all intermediate outputs for debugging.

    Args:
        row (dict): Dictionary containing:
            - "responses" (list): List of generated responses.
            - "grounding_llama32_70b_processed" (list): Corresponding "Yes" or "No" groundings.
            - "question" (str): The original question prompting these responses.
        tokenizer: Pre-trained tokenizer.
        model: Pre-trained entailment model.
        device (str): "cpu" or "cuda".

    Returns:
        float: Final grounding score with entropy.
    """

    # Parse responses & groundings from row data
    responses = ast.literal_eval(row["responses"])
    groundings = ast.literal_eval(row["grounding_llama32_70b_processed"])
    question = row["question"]
    n = len(responses)

    # Handle edge case: No responses or mismatched groundings
    if n == 0 or len(groundings) != n:
        print("No valid responses or mismatch between responses and groundings.")
        return 0.0

    print(f"\n--- Question: {question} ---")
    print("Responses:", responses)
    print("Groundings:", groundings)

    # ---- STEP 1: Build Adjacency Matrix for Semantic Clustering ----
    adjacency = [[False] * n for _ in range(n)]

    print("\nBuilding adjacency matrix using entailment...")

    for i in range(n):
        for j in range(i + 1, n):
            if are_semantically_equivalent(responses[i], responses[j], question, tokenizer, model, device=device):
                adjacency[i][j] = adjacency[j][i] = True

    print("Adjacency Matrix:")
    for row in adjacency:
        print(row)

    # ---- STEP 2: Cluster Responses Using BFS ----
    visited = [False] * n
    clusters = []

    print("\nClustering responses using BFS...")

    for i in range(n):
        if not visited[i]:
            stack = [i]
            visited[i] = True
            cluster_members = []

            while stack:
                node = stack.pop()
                cluster_members.append(node)
                for nb in range(n):
                    if adjacency[node][nb] and not visited[nb]:
                        visited[nb] = True
                        stack.append(nb)

            clusters.append(cluster_members)

    print("\nClusters formed:")
    for idx, cluster in enumerate(clusters):
        print(f"Cluster {idx+1}: {[responses[i] for i in cluster]}")

    # ---- STEP 3: Compute Grounding Entropy per Cluster ----
    grounding_scores_with_entropy = []

    print("\nComputing grounding entropy per cluster...")

    for idx, cluster in enumerate(clusters):
        cluster_size = len(cluster)
        yes_count = sum(1 for i in cluster if groundings[i].lower() == "yes")

        # Compute probability of "Yes" in cluster
        p = yes_count / cluster_size if cluster_size > 0 else 0

        # Compute Shannon entropy, handling edge cases
        if p in [0, 1]:  # No uncertainty in cluster
            h_cluster = 0.0
        else:
            h_cluster = - (p * math.log2(p) + (1 - p) * math.log2(1 - p))

        # Compute grounding score (fraction of "Yes" responses)
        grounding_score = yes_count / cluster_size if cluster_size > 0 else 0

        # Compute uncertainty-weighted grounding score
        grounding_score_with_entropy = grounding_score * h_cluster
        grounding_scores_with_entropy.append(grounding_score_with_entropy)

        # Print debug information
        print(f"\nCluster {idx+1}:")
        print(f"  Responses: {[responses[i] for i in cluster]}")
        print(f"  Groundings: {[groundings[i] for i in cluster]}")
        print(f"  Yes Count: {yes_count}, Total: {cluster_size}, P(Yes): {p:.4f}")
        print(f"  Entropy: {h_cluster:.4f}")
        print(f"  Grounding Score: {grounding_score:.4f}")
        print(f"  Weighted Grounding Score: {grounding_score_with_entropy:.4f}")

    # Compute final score using weighted sum
    final_score = sum(grounding_scores_with_entropy) / len(grounding_scores_with_entropy) if grounding_scores_with_entropy else 0.0

    print("\nFinal Grounding Score with Entropy:", final_score)
    return final_score
```

```python
def calc_external_grounding_uncertainty(row):
    """
    Computes uncertainty score based on multiple grounding models.
    """

    try:
        modelA = ast.literal_eval(row["grounding_llama32_11b_processed_score"])
        modelB = ast.literal_eval(row["grounding_llama32_70b_processed_score"])
        modelC = ast.literal_eval(row["grounding_qwen_vl_processed_score"])
    except:
        modelA = row["grounding_llama32_11b_processed_score"]
        modelB = row["grounding_llama32_70b_processed_score"]
        modelC = row["grounding_qwen_vl_processed_score"]

    if not isinstance(modelA, list) or not isinstance(modelB, list) or not isinstance(modelC, list):
        return 0.0  # Ensure inputs are lists

    num_responses = min(len(modelA), len(modelB), len(modelC))

    if num_responses == 0:
        return 0.0  # No responses

    total_score = 0.0
    for i in range(num_responses):
        answers = [
            str(modelA[i]).lower().strip(),
            str(modelB[i]).lower().strip(),
            str(modelC[i]).lower().strip()
        ]
        yes_count = sum(ans == "yes" for ans in answers)
        p = yes_count / 3.0

        H = 0.0 if p in [0, 1] else - (p * math.log2(p) + (1 - p) * math.log2(1 - p))
        resp_score = H * p
        total_score += resp_score

    return total_score / num_responses
```

```python
def load_grounding_dict(grounding_folder, score_key='biomedclip_score', score_type='binary'):
    out = {}
    out_processed = {}
    if not os.path.isdir(grounding_folder):
        print(f"[WARNING] Missing folder: {grounding_folder}")
        return out, out_processed

    pkls = [p for p in os.listdir(grounding_folder) if p.endswith('.pkl')]
    for pklf in pkls:
        fpath = os.path.join(grounding_folder, pklf)
        if os.path.getsize(fpath) == 0:
            continue
        with open(fpath, 'rb') as f:
            data = pickle.load(f)
        # print('qid',data.get('question_id', None))
        qid = int(data.get('question_id', None))
        val_list = []
        for key in data.keys():
            if 'response' in key:
                val_list.append(data[key].get(score_key, "No"))

        if qid:
            out[qid] = val_list
            if score_type == 'binary':
                temp = ['yes' if "yes" in v.lower() else 'no' for v in val_list]
                out_processed[qid] = temp
            elif score_type == 'continuous':
                temp = [float(v) if isinstance(v, (int, float)) else 0.0 for v in val_list]
                out_processed[qid] = temp
            else:
                print(f"[WARNING] Unknown score_type: {score_type}")
                out_processed[qid] = val_list
    return out, out_processed
```

```python
import numpy as np
import matplotlib.pyplot as plt


def plot_multiple_calibration_curve(df, confidence_cols, correctness_col, nbins=10, plot_title='temp', save_path=None):
    """
    Plots multiple reliability diagrams (confidence vs. accuracy) on the same plot, calculates ECE for each,
    and optionally saves the figure to save_path.

    Parameters
    ----------
    df : pandas.DataFrame
        The dataframe that contains the confidence and correctness columns.
    confidence_cols : [str]
        This is a list of columns with predicted probabilities (floats between 0 and 1).
    correctness_col : str
        Name of the column indicating correctness (1 for correct, 0 for incorrect).
    nbins : int
        Number of bins to group confidences.
    save_path : str or None
        If provided, will save the plot to this file path. Otherwise, calls plt.show().
    """

    fig, ax = plt.subplots(figsize=(8, 6))

    for confidence_col in confidence_cols:
        confidences = df[confidence_col].values
        correctness = df[correctness_col].values

        # Binning: We create evenly spaced bins from 0 to 1
        bins = np.linspace(0, 1, nbins + 1)  # e.g. if nbins=10, bins are [0, 0.1, 0.2, ... 1.0]
        bin_indices = np.digitize(confidences, bins) - 1  # Each confidence mapped to a bin index [0..nbins-1]

        # Precompute sums/counts in each bin
        bin_conf_sums = np.zeros(nbins, dtype=float)
        bin_acc_sums = np.zeros(nbins, dtype=float)
        bin_counts = np.zeros(nbins, dtype=int)

        for i in range(len(confidences)):
            b = bin_indices[i]
            if 0 <= b < nbins:
                bin_conf_sums[b] += confidences[i]
                bin_acc_sums[b] += correctness[i]
                bin_counts[b] += 1

        # Now compute mean confidence & mean accuracy for each bin
        bin_conf_means = []
        bin_acc_means = []
        bin_centers = []

        for b in range(nbins):
            if bin_counts[b] > 0:
                mean_conf = bin_conf_sums[b] / bin_counts[b]
                mean_acc = bin_acc_sums[b] / bin_counts[b]

                # For plotting, let's use the midpoint of that bin (optional)
                center = 0.5 * (bins[b] + bins[b + 1])

                bin_conf_means.append(mean_conf)
                bin_acc_means.append(mean_acc)
                bin_centers.append(center)

        # Calculate ECE (weighted by bin size)
        total_samples = len(confidences)
        ece = 0.0
        for b in range(nbins):
            if bin_counts[b] > 0:
                mean_conf = bin_conf_sums[b] / bin_counts[b]
                mean_acc = bin_acc_sums[b] / bin_counts[b]
                fraction = bin_counts[b] / total_samples
                ece += fraction * abs(mean_acc - mean_conf)

        # Plot the reliability diagram for this confidence column
        ax.plot(bin_conf_means, bin_acc_means, marker='o', linestyle='-', linewidth=2, label=f'{confidence_col} (ECE = {ece:.3f})')

    ax.plot([0, 1], [0, 1], '--', color='gray', label='Perfect Calibration')

    ax.set_xlim([0, 1])
    ax.set_ylim([0, 1])
    ax.set_xlabel('Confidence', fontsize=14)
    ax.set_ylabel('Accuracy', fontsize=14)
    ax.set_title(f'{plot_title} Reliability Diagram', fontsize=16)
    ax.legend(loc='best')
    plt.show()
```
