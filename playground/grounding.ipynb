```python
# import matplotlib.pyplot as plt
# import seaborn as sns

# # Use the same style and scale
# sns.set(style="whitegrid", font_scale=1.4, rc={'axes.labelweight': 'bold', 'axes.titleweight': 'bold'})
# plt.figure(figsize=(5,5))  # slightly smaller to match the left plot's layout

# # Match the color and aesthetics more closely
# sns.histplot(
#     df2['accuracy'],
#     bins=20,
#     kde=True,
#     color='#cd5c5c',         # darker pink (like light coral)
#     edgecolor='black',       # bar edges
#     linewidth=1
# )

# # Bold title and labels
# plt.title('Histogram of Accuracy', fontsize=14, fontweight='bold')
# plt.xlabel('Accuracy', fontsize=13, fontweight='bold')
# plt.ylabel('Frequency', fontsize=13, fontweight='bold')

# # Bold tick labels
# plt.xticks(fontsize=12, fontweight='bold')
# plt.yticks(fontsize=12, fontweight='bold')

# plt.tight_layout()
# plt.show()

import matplotlib.pyplot as plt
import seaborn as sns

# Set aesthetic style and scale
sns.set(
    style="whitegrid", 
    context="notebook", 
    font_scale=1.3,
    rc={
        'axes.labelweight': 'bold', 
        'axes.titleweight': 'bold',
        'axes.edgecolor': 'gray',
        'grid.alpha': 0.3,
        'grid.linestyle': '--'
    }
)

# Create the plot
plt.figure(figsize=(6, 5))

sns.histplot(
    df['accuracy'],
    bins=20,
    kde=True,
    color='#CD5C5C',          # Coral color
    edgecolor='black',
    linewidth=0.8
)

# Add title and axis labels with consistent styling
plt.title("LLaVA-Med's Accuracy  on Slake", fontsize=20, fontweight='bold', pad=15)
plt.xlabel('Accuracy', fontsize=20, fontweight='bold', labelpad=10)
plt.ylabel('Frequency', fontsize=20, fontweight='bold', labelpad=10)

# Customize tick labels
plt.xticks(fontsize=12, fontweight='bold')
plt.yticks(fontsize=12, fontweight='bold')

# Add tight layout and subtle grid
plt.grid(True, linestyle='--', linewidth=0.5, alpha=0.6)

plt.tight_layout()
plt.show()
```
